## Making sure exact video we are extracting from pool of original video

```python
import csv
import os

train_csv = r"D:\FYP\MSSG\data\AVA\csv\train_orig.csv"
val_csv   = r"D:\FYP\MSSG\data\AVA\csv\val_orig.csv"

def extract_video_ids(csv_path):
    video_ids = set()
    
    with open(csv_path, newline='', encoding='utf-8') as f:
        reader = csv.DictReader(f)
        for row in reader:
            video_ids.add(row['video_id'])
    
    return video_ids

# Extract
train_ids = extract_video_ids(train_csv)
val_ids   = extract_video_ids(val_csv)

# Print results
print("=== TRAIN VIDEO IDS ===")
for vid in sorted(train_ids):
    print(vid)

print("\n=== VAL VIDEO IDS ===")
for vid in sorted(val_ids):
    print(vid)

# Optional: show counts
print("\nSummary:")
print(f"Train count: {len(train_ids)}")
print(f"Val count:   {len(val_ids)}")
```

## Check clips video and audio extracted actually contains in pool

```python
import os

audio_dir = r"D:\FYP\MSSG\data\AVA\orig_audios\trainval"
video_dir = r"D:\FYP\MSSG\data\AVA\orig_videos\trainval"

def check_files(video_ids, split_name):
    missing_audio = []
    missing_video = []
    missing_both = []
    
    for vid in video_ids:
        audio_path = os.path.join(audio_dir, f"{vid}.wav")
        video_path = os.path.join(video_dir, f"{vid}.mp4")
        
        audio_exists = os.path.exists(audio_path)
        video_exists = os.path.exists(video_path)
        
        if not audio_exists and not video_exists:
            missing_both.append(vid)
        elif not audio_exists:
            missing_audio.append(vid)
        elif not video_exists:
            missing_video.append(vid)
    
    print(f"\n=== {split_name} CHECK ===")
    print(f"Total videos: {len(video_ids)}")
    print(f"Missing audio only: {len(missing_audio)}")
    print(f"Missing video only: {len(missing_video)}")
    print(f"Missing BOTH: {len(missing_both)}")
    
    return missing_audio, missing_video, missing_both

# Run checks
train_missing_audio, train_missing_video, train_missing_both = check_files(train_ids, "TRAIN")
val_missing_audio, val_missing_video, val_missing_both = check_files(val_ids, "VAL")

print("\n=== TRAIN missing BOTH ===")
for vid in train_missing_both:
    print(vid)

print("\n=== VAL missing BOTH ===")
for vid in val_missing_both:
    print(vid)
```

## Issue generation

```python
import os

# Path to your file list
file_list_path = r"D:\FYP\MSSG\predata\facebody_process\split\train_file_list.4"
# Root directory where folders should exist
video_root = r"D:\FYP\MSSG\data\AVA\clips_videos_face_body\train"

# Step 1: Extract video IDs from file list
expected_ids = set()

with open(file_list_path, "r") as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        
        # Example line:
        # ./predata/csv_split/train/_a9SWtcaNj8.csv
        video_id = os.path.basename(line).replace(".csv", "")
        expected_ids.add(video_id)

# Step 2: Get actual folder names in directory
existing_ids = set()

if os.path.exists(video_root):
    for name in os.listdir(video_root):
        full_path = os.path.join(video_root, name)
        if os.path.isdir(full_path):
            existing_ids.add(name)

# Step 3: Find missing IDs
missing_ids = expected_ids - existing_ids

# Step 4: Print results
print(f"Total expected IDs: {len(expected_ids)}")
print(f"Total existing IDs: {len(existing_ids)}")
print(f"Missing IDs: {len(missing_ids)}\n")

for vid in sorted(missing_ids):
    print(vid)
```
